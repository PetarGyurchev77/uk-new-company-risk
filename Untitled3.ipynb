```python
%pip install -q "psycopg[binary]"
from google.colab import userdata
import time
import datetime as dt

import requests
import psycopg
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

api_key = userdata.get("CH_API_KEY")
database_url = userdata.get("DATABASE_URL")

URL = "https://api.company-information.service.gov.uk/advanced-search/companies"
PAGE = 5000
MAX_RESULTS = 10000


session = requests.Session()
session.mount("https://", HTTPAdapter(max_retries=Retry(
    total=5,
    backoff_factor=2,
    status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"],
    respect_retry_after_header=True,
)))

```

```python
def fetch_day(day):
    """Return (hits, items) for one incorporation date, paging past 5,000 if needed."""
    params = {"incorporated_from": day, "incorporated_to": day, "size": PAGE}
    r = session.get(URL, params=params, auth=(api_key, ""), timeout=60)
    if r.status_code == 404:
        return 0, []
    r.raise_for_status()
    data = r.json()
    hits, items = data["hits"], data.get("items", [])

    start = PAGE
    while len(items) < hits and start < MAX_RESULTS:
        time.sleep(1)
        params["start_index"] = start
        r = session.get(URL, params=params, auth=(api_key, ""), timeout=60)
        r.raise_for_status()
        page = r.json().get("items", [])
        r = session.get(URL, params=params, auth=(api_key, ""), timeout=60)


    if hits > MAX_RESULTS:
        print(f"  WARNING {day}: {hits} hits, but the API only returns the first {MAX_RESULTS}")
    return hits, items



SQL_COMPANY = """
    INSERT INTO raw_companies (
        company_number, company_status, company_name, company_type,
        date_of_creation, date_of_cessation, country, address_line_1,
        address_line_2, locality, region, postal_code)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    ON CONFLICT DO NOTHING
"""
SQL_SIC = """
    INSERT INTO raw_company_sic (company_number, sic_code)
    VALUES (%s, %s)
    ON CONFLICT DO NOTHING
"""

SQL_LOG = """
    INSERT INTO raw_pull_log (incorporation_date, hits_reported, rows_received)
    VALUES (%s, %s, %s)
"""
SQL_DONE = "SELECT incorporation_date FROM raw_pull_log"


def company_row(item):
    """Turn one API item into a tuple, in the same order as the columns in SQL_COMPANY."""
    addr = item.get("registered_office_address", {})
    return (
        item["company_number"],
        item.get("company_status"),
        item.get("company_name"),
        item.get("company_type"),
        item.get("date_of_creation"),
        item.get("date_of_cessation"),
        addr.get("country"),
        addr.get("address_line_1"),
        addr.get("address_line_2"),
        addr.get("locality"),
        addr.get("region"),
        addr.get("postal_code"),
    )


def pull_day(conn, day):
    """Pull one day and load it. All three tables commit together, or not at all."""
    hits, items = fetch_day(day)
    company_rows = [company_row(i) for i in items]
    sic_rows = [(i["company_number"], code) for i in items for code in i.get("sic_codes", [])]
    with conn.cursor() as cur:
        cur.executemany(SQL_COMPANY, company_rows)
        cur.executemany(SQL_SIC, sic_rows)
        cur.execute(SQL_LOG, (day, hits, len(items)))
    conn.commit()
    return hits, len(items)


def pull_range(start, end):
    """Pull every day from start to end inclusive. Safe to stop and re-run at any time."""
    with psycopg.connect(database_url) as conn:
        done = {row[0] for row in conn.execute(SQL_DONE).fetchall()}
        conn.commit()
        days = [start + dt.timedelta(days=n) for n in range((end - start).days + 1)]
        todo = [d for d in days if d not in done]
        print(f"{len(days)} days in range, {len(days) - len(todo)} already pulled, {len(todo)} to go")

        t0, failed = time.time(), []
        for n, day in enumerate(todo, 1):
            try:
                hits, received = pull_day(conn, day.isoformat())
                flag = "" if hits == received else "  <-- MISMATCH"
                elapsed = time.time() - t0
                eta = elapsed / n * (len(todo) - n) / 60
                print(f"[{n}/{len(todo)}] {day}  hits {hits:>5}  received {received:>5}  ~{eta:.0f} min left{flag}")
            except Exception as e:
                conn.rollback()          #
                failed.append(day)
                print(f"[{n}/{len(todo)}] {day}  FAILED: {e!r}")
            time.sleep(1)

        print(f"\nDone in {(time.time() - t0) / 60:.1f} min. {len(failed)} failed day(s): {failed}")
        if failed:
            print("Run pull_range again with the same dates: it will retry only those days.")
```
